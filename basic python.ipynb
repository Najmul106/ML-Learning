```python
a_lst = [4, 2, 7, 2, 4, 9, 1]

def first():
    sorted_lst = sorted(a_lst)
    unique_lst = list(set(sorted_lst))
    if unique_lst:
        unique_lst.remove(min(unique_lst))
    
    return sorted_lst, list(set(sorted_lst)), unique_lst

a, b, c = first()
print(a)
print(b)
print(c)
```

```python
def count_word_frequencies(text):
    words = text.lower().split()
    freq_dict = {}   
    for word in words:
        freq_dict[word] = freq_dict.get(word, 0) + 1    
    return freq_dict
input_text = "Apple banana apple orange Banana orange"
output = count_word_frequencies(input_text)
print(output)
```
